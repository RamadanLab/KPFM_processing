```python
# Adjust based on analysis wanted. 
def full_analysis(file_directory, filetype, channel_type, 
                  image_time = None, cutoff1 = None, cutoff2 = None, 
                  plotname = None, color = None, measurement = '', scale = '', load_files = ''):
    f1, f2, f3, f4 = AFMRead.setup_directories(file_directory, raw_folder_name="raw_data")
    if load_files == "yes": 
        AFMRead.load_from_folder(folder_path = f1, save_path = f2, filetype = filetype, channel_type = channel_type)
    else: 
        print("npy files ready")
    gaussian_peaks1, time1, x_left1, x_right1 = tp.fitting_folder(f2, image_time, f3, measurement)

    pf.time_series_plot(gaussian_peaks1, time1, x_left1, x_right1, 
                    cutoff_line1=cutoff1, cutoff_line2 = cutoff2,
                    save_path = f4, savename = plotname, 
                    color =color, measurement = measurement, 
                     scale = scale)
    return 
```
